```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from time import perf_counter
import numpy as np
```

```python
# Implement baseline algorithm
def baseline_max_sublist(arr):
    max_sum = float("-inf")
    for i in range(len(arr)):
        current_sum = 0
        for h in range(i, len(arr)):
            current_sum += arr[h]
            if current_sum > max_sum:
                max_sum = current_sum

    return max_sum
```

```python
# Implement target algorithm
def KadaneMaxSubarray(A): 
    max_so_far = float("-inf")
    current_max = 0
    for i in range(len(A)):
        current_max = current_max + A[i]
        if current_max > max_so_far:
           max_so_far = current_max
        if current_max < 0:
           current_max = 0
    return max_so_far
```

```python
def random(n):


    return np.random.randint(-100, 101, size=n)
```

```python
testing = [-3, -1, 1, 5, 3, 4]
print("baseline:", baseline_max_sublist(testing))
print("kadan:", KadaneMaxSubarray(testing))

assert baseline_max_sublist(testing) == 9
assert KadaneMaxSubarray(testing) == 9

neg = [-3, -1, -5, -3, -4]

assert baseline_max_sublist(neg) == -1
assert KadaneMaxSubarray(neg) == -1
```

```python
sizes = [100, 500, 1000, 2500, 5000, 10000]
baseline_times = []
kadane_times = []
for n in sizes:
    baseline_total = 0
    kadane_total = 0
    for j in range(5):
        arr = random(n)
        start = perf_counter()
        baseline_max_sublist(arr)
        baseline_total += perf_counter() - start

        start = perf_counter()
        KadaneMaxSubarray(arr)
        kadane_total += perf_counter() - start
    baseline_times.append(baseline_total / 5)
    kadane_times.append(kadane_total / 5)
for i in range(len(sizes)):
    print("N =", sizes[i], "baseline =", baseline_times[i],"kadane =", kadane_times[i])
```

```python
plt.figure(figsize=(8,5))
plt.plot(sizes, baseline_times, marker="o", label="Baseline")
plt.plot(sizes, kadane_times, marker="o", label="kadane")
plt.xlabel("array size (n)")
plt.ylabel("Time (seconds)")
plt.title("Baseline vs Kadane")
plt.legend()
plt.grid(True)
plt.show()
```
